```python
import pandas as pd

df = pd.read_csv("amazon.csv")
df.head()
```

```python
df.info()
df.isnull().sum()
df.duplicated().sum()
```

1. product_id must be unique
2. price must be > 0
3. rating must be between 0–5
4. product_name must not be NULL
5. no duplicate rows

```python
import pandas as pd

# price clean: ₹ and comma remove
df['discounted_price_num'] = (
    df['discounted_price'].astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['actual_price_num'] = (
    df['actual_price'].astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['discounted_price_num'] = pd.to_numeric(df['discounted_price_num'], errors='coerce')
df['actual_price_num'] = pd.to_numeric(df['actual_price_num'], errors='coerce')

# rating clean
df['rating_num'] = pd.to_numeric(df['rating'], errors='coerce')

# rating_count clean
df['rating_count_num'] = (
    df['rating_count'].astype(str)
    .str.replace(',', '', regex=False)
)
df['rating_count_num'] = pd.to_numeric(df['rating_count_num'], errors='coerce')

df[['discounted_price','discounted_price_num','actual_price','actual_price_num','rating','rating_num','rating_count','rating_count_num']].head()
```

```python
def validate(df):
    errors = []

    if df.duplicated().sum() > 0:
        errors.append("❌ Duplicate rows found")

    if df['product_id'].duplicated().sum() > 0:
        errors.append("❌ Duplicate product_id found")

    if df['product_name'].isnull().sum() > 0:
        errors.append("❌ product_name has null values")

    if not (df['discounted_price_num'] > 0).all():
        errors.append("❌ discounted_price has invalid values")

    if not (df['actual_price_num'] > 0).all():
        errors.append("❌ actual_price has invalid values")

    if not ((df['rating_num'] >= 0) & (df['rating_num'] <= 5)).all():
        errors.append("❌ rating out of range")

    if not (df['rating_count_num'] >= 0).all():
        errors.append("❌ rating_count invalid")

    if not (df['discounted_price_num'] <= df['actual_price_num']).all():
        errors.append("❌ discounted_price greater than actual_price")

    return errors

issues = validate(df)

if issues:
    print("Data Issues:")
    for i in issues:
        print(i)
else:
    print("✅ Data Clean")
```

```python
# 1) rating numeric clean first (if not already done)
df = df[df['rating_num'].notnull()]
df = df[(df['rating_num'] >= 0) & (df['rating_num'] <= 5)]

# 2) rating_count clean
df['rating_count_num'] = (
    df['rating_count'].astype(str)
    .str.replace(',', '', regex=False)
    .str.extract(r'(\d+)')
)
df['rating_count_num'] = pd.to_numeric(df['rating_count_num'], errors='coerce')
df = df[df['rating_count_num'].notnull()]

# 3) duplicates (ONLY once)
df = df.drop_duplicates(subset=['product_id'], keep='first')
```

```python
issues = validate(df)

if issues:
    print("Data Issues:")
    for i in issues:
        print(i)
else:
    print("✅ Data Clean")
```
